### O que é o Avonet?

O banco de dados AVONET contém dados abrangentes sobre características funcionais de todas as aves, incluindo seis variáveis ​​ecológicas, onze características morfológicas contínuas e informações sobre a extensão e a localização de suas áreas de ocorrência. Medições morfológicas brutas estão disponíveis para 90.020 indivíduos, abrangendo 11.009 espécies de aves existentes, amostradas em 181 países. Esses dados também são apresentados na forma de médias por espécie, organizadas em três formatos taxonômicos, o que permite a integração com uma filogenia global, mapas de distribuição geográfica, dados da Lista Vermelha da IUCN e a base de dados de ciência cidadã eBird. Também está incluído o código necessário para reproduzir as análises e figuras apresentadas no artigo de Tobias et al. (2022), intitulado "AVONET: morphological, ecological and geographical data for all birds", publicado na revista *Ecology Letters*.

## O que realmente pode ser utilizado?

Acabamos de ver que temos muitas informações interessantes na nossa tabela. Naturalmente, o próximo passo seria iniciar a limpeza dos dados, começando com a remoção de linhas que tivessem dados faltantes. Porém, algo inesperado acontece quando removemos as linhas que contenham campos faltantes, com o uso do método `.dropna()`:

```python
df_original.dropna()
```

Todas as linhas do dataset são removidas quando removemos as linhas que contém dados faltantes, ou seja, não existe nenhuma linha que tenha todas as informações registradas. Como não podemos fazer nada com uma tabela vazia, precisamos começar a escolher quais atributos são interessantes para nosso objetivo, e quais atributos podem ser ignorados.

Considerando que temos um grave problema de dados faltantes, vamos percorrer todas as colunas, somar todas as instâncias de `Nan` em cada uma delas e exibir o resultado. Assim, podemos saber quantos dados faltantes existem em cada uma das colunas, e consequentemente, podemos saber quais colunas têm mais dados faltando e quais colunas têm menos dados faltando.

```python
valores_ausentes = {}

for coluna in df_original.columns:
  contagem = df_original[coluna].isna().sum()
  total_registros = len(df_original[coluna])

  print(f"Coluna analisada: {coluna}")
  print(f"Número de valores faltantes nessa coluna: {contagem}")
  print(f"Número de valores preenchidos nessa coluna: {total_registros - contagem}")
  print("------")

  valores_ausentes[coluna] = contagem
```

Também vamos apresentar esses resultados em um gráfico de barras, para observarmos melhor:

```python
eixo = sns.barplot(
    valores_ausentes,
    x = valores_ausentes.values(),
    y = valores_ausentes.keys(),
    palette = "viridis",
    hue = list(valores_ausentes.values())
)

plt.xlabel("Número de valores ausentes")
plt.ylabel("Coluna")
plt.title("Valores faltantes por coluna")
```

Com base nessas informações, podemos perceber algumas coisas bem importantes:

A coluna `publicacao` tem 87.062 valores ausentes, com apenas 3.309 linhas preenchidas. Sendo assim, essas três colunas serão removidas, justamente pela quantidade de dados faltantes.

As colunas   `indice_asa_mao`, `distancia_kipp` e `comprimento_pena_secundaria`, em comparação com as colunas observadas anteriormente, não apresentam tantos dados ausentes assim. Porém, as três colunas apresentam mais de 26.000 dados ausentes, e portanto, também serão removidas. A coluna de `numero_individuo` também será removida, visto que apresenta 16.578 dados faltando, e que ela não será um atributo relevante em futuras predições.

As colunas `especie1_BirdLife`, `especie2_eBird` e `especie3_BirdTree` são praticamente a mesma coisa, e não faria sentido incluí-las nos modelos futuros. Sendo assim, considerando que a catalogação do BirdLife é a principal referência para catalogação de espécies do AVONET, vamos remover as outras duas.

Apesar das colunas de `data_type`, `source`, `medidor` e `protocolo` não serem colunas que serão muito relevante nos modelos que serão futuramente desenvolvidos, vamos mantê-las na tabela, já que nenhum deles têm um volume de dados ausentes muito relevante - nenhum deles passa das 400 ausências.


```python
colunas_REMOVIDAS = [
  'localizacao',
  'pais',
  'publicacao',
  'protocolo',
  'numero_individuo',

  'indice_asa_mao',
  'distancia_kipp',
  'comprimento_pena_secundaria',

  'especie2_eBird',
  'especie3_BirdTree'
]

df_novo = df_original.copy()
df_novo = df_novo.drop(columns = colunas_REMOVIDAS).dropna()
df_novo
```

Podmeos observar que depois de removermos as colunas que mais tinham dados faltando e outras colunas que não eram necessárias, ficamos com 45.205 linhas, que é quase metade do que tinhamos originalmente.

A próxima etapa será filtrarmos o dataset para que fiquem apenas indivíduos cujas espécies possuam, pelo menos, 50 registros no dataset. Para isso, vamos contar quantos exemplares cada espécie tem, com o método `.value_counts()`, e em seguida, filtrar as espécies que tenham ao menos 50 ocorrências. Sempre que encontrarmos uma espécie que atenda ao requisito, essa espécie será adicionada na lista de espécies desejadas.

```python
contagem = df_novo["especie1_BirdLife"].value_counts()
especies_desejadas = []

for especie, quantidade in contagem.items():
  if quantidade > 50:
    print(f"Espécie: {especie}")
    print(f"Número de exemplares: {quantidade}")
    print("------")
    especies_desejadas.append(especie)
```

Com isso, chegamos a oito espécies que atendem ao nosso requisito, sendo elas:
- Diglossa brunneiventris
- Mionectes striaticollis
- Cyanomitra olivacea
- Eurillas virens
- Atlapetes rufinucha
- Elaenia pallatangae
- Diglossa cyanea
- Phyllastrephus cabanisi

Agora, como queremos trabalhar só com essas espécies, vamos filtrar o nosso dataframe para que ele inclua apenas as espécies que selecionamos:

```python
df_filtrado = df_novo.copy()

logica = df_filtrado["especie1_BirdLife"].isin(especies_desejadas)
df_filtrado = df_filtrado[logica]
```

Por fim, sabemos que o python tem alguns históricos problemáticos com dados numéricos e dados de string. Sendo assim, para garantir que os dados numéricos sejam, de fato, numéricos, vamos transformar todos os dados das colunas numéricas em numéros.

Para isso, criamos uma lista que contenha todas as colunas numéricas, e em seguida, percorremos essa lista. A partir disso, para cada elemento dessa lista, utilizamos o método `.to_numeric()` do pandas para converter os dados para numéricos. Apesar de já termos removidas os dados `NaN` anteriormente, utilizamos o método `.fillna(0)` para assegurar que não teremos nenhum dado faltante em nosso conjunto de dados.

### Criando dataframes individuais para cada espécie

A fim de tornar a análise das espécies individuais mais fácil, vamos criar dataframes individuais para cada uma dessas espécies, separando suas informações.

Para isso, definimos a função `cria_dataframe_individual_para_especie()`, que recebe a espécie que terá um dataframe individual, copia o dataframe original e filtra para apenas os registros da espécie desejada, retornando o dataframe criado.

```python
def cria_dataframe_individual_para_especie(especie, dataframe = df_filtrado):
    """Cria um dataframe separado para uma espécie em específico, com base no dataframe original"""

    novo_df = dataframe.copy()
    logica = dataframe["especie1_BirdLife"] == especie
    novo_df = novo_df[logica]

    return novo_df
```

Com a definição da função, vamos aplicá-la para todas as espécies que separamos anteriormente, criando um dataframe individual para cada uma delas:

```python
df_brunneiventris = cria_dataframe_individual_para_especie("Diglossa brunneiventris")
df_striaticollis = cria_dataframe_individual_para_especie("Mionectes striaticollis")
df_olivacea = cria_dataframe_individual_para_especie("Cyanomitra olivacea")
df_virens = cria_dataframe_individual_para_especie("Eurillas virens")
df_rufinucha = cria_dataframe_individual_para_especie("Atlapetes rufinucha")
df_pallatangae = cria_dataframe_individual_para_especie("Elaenia pallatangae")
df_cyanea = cria_dataframe_individual_para_especie("Diglossa cyanea")
df_cabanisi = cria_dataframe_individual_para_especie("Phyllastrephus cabanisi")
```

### Dados numéricos - Análises

Para termos uma melhor noção das medidas que estamos trabalhando, vamos fazer algumas medidas dos nossos dados numéricos, vamos realizar algumas medidas para termos uma noção melhor das dimensões de cada espécie. Para isso, vamos utilizar o método `.describe()`, do pandas, que nos fornece dados como média, desvio padrão, mínimo, máximo, dentre outros. É importante ressaltar que o método do `.describe()` se aplica para dados que tenham forma de número, embora ele não faça distinção entre dados nominais e ordinais, o que discutiremos assim que aparecer pela primeira vez.

#### Funções importantes

```python
colunas = [
    'comprimento_bico_culmen',
    'comprimento_bico_nares',
    'largura_bico',
    'profundidade_bico',
    'comprimento_tarso',
    'comprimento_asa',
    'comprimento_cauda'
]

def plotar_distribuicao_continua(dataframe, coluna, pos_painel, unidade_medida = "mm", cor = 'navy'):
    """Plota um gráfico de distribuição contínua de uma coluna de um dataframe"""

    titulo = "Função densidade de probabilidade para " + coluna

    #plota distribuição geral e sua média
    eixo = sns.kdeplot(
      data = dataframe,
      x = coluna,
      color = cor,
      ax = pos_painel,
      label = ("Distribuição: Geral")
    )

    # linha de média
    eixo.axvline(
          st.mean( dataframe[coluna]),
          linestyle = "dotted",
          color = "red",
          label = ("Média geral: " + str(round((( dataframe[coluna]).mean()), 2))) + unidade_medida,
    )


    # coloca gradezinha de fundo, legendas e titulo
    eixo.grid(True, linestyle = "dotted", alpha = 0.8)
    eixo.set_xlabel(coluna + " (" + unidade_medida + ")")
    eixo.set_title( titulo )
    eixo.legend()

    plt.tight_layout(pad=3.0)
```

```python
def conjunto_graficos(dataframe, colunas, n_colunas = 3, n_linhas = 3, cor = 'navy'):
  fig, axes = plt.subplots(
      nrows = n_linhas,
      ncols = n_colunas,
      figsize = (18, 14),
      )

  fig.suptitle('Distribuição de medidas da espécie ' + dataframe['especie1_BirdLife'].iloc[0], fontsize = 16)

  i = 0
  j = 0
  for coluna in colunas:
    plotar_distribuicao_continua(dataframe, coluna, axes[i, j], cor = cor)
    j += 1

    if j == n_colunas:
      j = 0
      i += 1

  axes[2, 1].axis('off')
  axes[2, 2].axis('off')
```

### Box plot

#### Funçãozinha

```python
def plotar_grafico_scatterplot_duplo(dataframe, eixoX, eixoY, titulo, palette, pos_legenda = 2, fs = (13, 8)):
  """Plota um gráfico scatterplot da distribuição entre o eixoX e o eixoY"""
  plt.figure(figsize=fs)

  # conta os pontos de mesma posiçõa e especie, para apresentarmos a densidade dos pontos no gráfico
  contagens = (
      dataframe.groupby([eixoX, eixoY, 'especie1_BirdLife'])
      .size()
      .reset_index(name='contagem')
  )

  contagens = contagens.sort_values(by='especie1_BirdLife')

  if palette is list:
      eixo = sns.scatterplot(
        data=contagens,
        x = eixoX,
        y = eixoY,
        hue = 'especie1_BirdLife',
        size = "contagem",
        #style = "especie1_BirdLife",

        # https://stackoverflow.com/questions/59695334/custom-color-palette-in-seaborn
        palette=sns.color_palette(palette, len(palette)),
        sizes=(80, 200),
        alpha = 0.9
      )

  else:
    eixo = sns.scatterplot(
        data=contagens,
        x = eixoX,
        y = eixoY,
        size = "contagem",
        color = palette,
        sizes=(80, 200),
        alpha = 0.9
      )

  # Legendas e títulos
  eixo.set_title(titulo)
  eixo.set_xlabel(eixoX)
  eixo.set_ylabel(eixoY)

  # Ajusta a posição da legenda com base no parâmetro pos_legenda
  sns.move_legend(eixo, loc=pos_legenda)
```

#### Funçãozinha

```python
def plotar_grafico_boxplot(dataframe, variavel, palette, pos_legenda = 2):
  """Plota um gráfico scatterplot da distribuição entre o eixoX e o eixoY"""
  titulo = "Box plot de " + variavel
  plt.figure(figsize=(16, 8))

  # plota o gráfico
  eixo = sns.boxplot(
    data=dataframe,
    x = dataframe['especie1_BirdLife'],
    y = variavel,
    palette = palette
  )

  # Legendas e títulos
  eixo.set_title(titulo)
  eixo.set_xlabel("Espécie")
  eixo.set_ylabel(variavel + " (mm)")
```
